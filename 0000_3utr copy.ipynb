Here is the exact Python code to extract 3'UTR features from a GTF database using gffutils and save them in BED format:

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import gffutils
from pathlib import Path
import os

# Set the desired path
new_path = "/Users/alejandraescos/Documents/github/SCR"
os.chdir(new_path)

# --- Paths ---
db_path = Path('grcm39.db')  # Adjust if needed
output_bed = Path('0005_riboseq/data/0002_utr_grcm39.bed')

# Ensure output directory exists
output_bed.parent.mkdir(parents=True, exist_ok=True)

# --- Load GTF database ---
db_gtf = gffutils.FeatureDB(db_path, keep_order=True)

# --- Extract 3' UTR features and write to BED ---
with open(output_bed, "w") as bed_out:
    for utr in db_gtf.features_of_type('three_prime_utr'):
        chrom = utr.chrom
        strand = utr.strand
        transcript_id = utr.attributes.get('transcript_id', ['unknown'])[0]
        gene_id = utr.attributes.get('gene_id', ['unknown'])[0]
        gene_symbol = utr.attributes.get('gene_name', [gene_id])[0]
        score = 0  # BED score field, set to 0

        # Shift start position by +15 nucleotides
        if strand == "+":
            start = utr.start - 1 + 15  # BED format uses 0-based start
            end = utr.end
        else:
            start = utr.start - 1
            end = utr.end - 15
        if end <= start:
            continue  # Skip invalid regions
        
        name_field = f"{gene_symbol}|{gene_id}|{transcript_id}"
        bed_line = f"{chrom}\t{start}\t{end}\t{name_field}\t{score}\t{strand}\n"
        bed_out.write(bed_line)

print(f"Saved shifted BED file to: {output_bed}")

```
